```python
import os
from crewai import Agent, Task, Crew, Process, LLM
from crewai.tools import tool
from google.colab import userdata
```

```python
os.environ["hashir_api-key"] = userdata.get("hashir_api-key")
# Restricting max_tokens and using the currently active qwen-2.5-coder-32b model
llm = LLM(
    model="groq/qwen-2.5-coder-32b",
    api_key=userdata.get("hashir_api-key"),
    max_tokens=300
)
print("LLM ready with active Qwen Coder model")
```

```python
from google.colab import files
uploaded = files.upload()
print("Uploaded:", list(uploaded.keys()))

from pypdf import PdfReader
```

```python
document_text = ""
for name in uploaded.keys():
    if name.lower().endswith(".pdf"):
        reader = PdfReader(name)
        for page in reader.pages:
            document_text += (page.extract_text() or "") + "\n"
    else:
        with open(name, "r", encoding="utf-8") as f:
            document_text += f.read() + "\n"

# quick sanity check so I know the text actually loaded
print(f"Total characters: {len(document_text)}")
print(document_text[:300])
```

```python
def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        piece = text[start:start + chunk_size].strip()
        if piece:
            chunks.append(piece)
        start += chunk_size - overlap
    return chunks

chunks = chunk_text(document_text)
print(f"Created {len(chunks)} chunks")
```

```python
!pip install "pillow>=10.4.0,<11.1.0" -q

from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")
chunk_embeddings = embedder.encode(chunks)
print("Embeddings shape:", chunk_embeddings.shape)
```

```python
import numpy as np

def retrieve(query, top_k=3):
    q = embedder.encode([query])[0]
    sims = (chunk_embeddings @ q) / (np.linalg.norm(chunk_embeddings, axis=1) * np.linalg.norm(q))
    best = np.argsort(sims)[::-1][:top_k]
    return [chunks[i] for i in best]

results = retrieve("What are the main purposes of customer relationship management?")
for r in results:
    print(r, "\n---")
```

```python
@tool("Document Search")
def document_search(query: str) -> str:
    """Searches the student's own uploaded documents for relevant passages."""
    results = retrieve(query, top_k=2)
    return "\n\n".join(r[:400] for r in results)   # short, so I stay under Groq's free token limit

researcher = Agent(
    role="Personal Notes Researcher",
    goal="Find the passages in the student's own notes that best answer the question, and report only what the notes actually say.",
    backstory=(
        "You are the student's study buddy who has read every page of their notes. "
        "You never go to the internet and you never guess. If the notes don't cover "
        "something, you say so plainly."
    ),
    tools=[document_search],
    llm=llm,
    max_iter=10,
)
```

```python
@tool("Fact Check Lookup")
def web_search(query: str) -> str:
    """Looks up a claim on the internet. Takes exactly one argument named 'query', a plain text search string, for example: 'main purposes of CRM'."""
    try:
        from ddgs import DDGS
        results = DDGS().text(str(query), max_results=2)
        return str(results)[:1500] or "No results found."
    except Exception as e:
        return f"Search unavailable right now ({type(e).__name__}). Say that the claim could not be verified."

verifier = Agent(
    role="Fact Checker",
    goal="Check whether the claims found in the student's notes are still correct according to current web sources.",
    backstory=(
        "You are a skeptical fact-checker. Your ONLY tool is 'Fact Check Lookup'. Never use web_open, "
        "browser tools, or any other tool. You run Fact Check Lookup once for the main claim, then decide "
        "if it holds up, and you say clearly what agrees and what doesn't."
    ),
    tools=[web_search],
    llm=llm,
    max_iter=10,
)
```

```python
writer = Agent(
    role="Answer Writer",
    goal="Turn the notes and the fact-check result into one clear, honest answer a student can actually use.",
    backstory=(
        "You are a clear, friendly writer. You have NO tools. Never call any function or tool; just write your "
        "answer in plain text from the notes and fact-check you are given. You explain things like a good tutor: "
        "short, plain, no fluff. If the notes and the web disagree, you say so."
    ),
    llm=llm,
    max_iter=10,
)

research_task = Task(
    description=(
        "Use the Document Search tool to find passages in the student's notes that answer this question: {question}\n"
        "Return the relevant passages and a one-line summary of what the notes claim. If the notes don't cover it, say so. "
        "Call the Document Search tool once or twice at most, then give your final answer."
    ),
    agent=researcher,
    expected_output="The relevant note passages plus a short summary of the notes' main claim(s).",
)

verify_task = Task(
    description=(
        "Take the main claim from the Researcher's findings about: {question}\n"
        "You MUST use the Fact Check Lookup tool to check that specific claim against current sources. "
        "Call it once or twice at most, with a plain text 'query' and nothing else. "
        "Report whether the web agrees, partly agrees, or contradicts the notes, then give your final answer. "
        "The only tool you may use is Fact Check Lookup. Never open web pages or use any other tool."
    ),
    agent=verifier,
    context=[research_task],
    expected_output="A verdict (agrees / partly agrees / contradicts) with a short explanation based on web search results.",
)

write_task = Task(
    description=(
        "Write the final answer to: {question}\n"
        "Use the notes and the verification result you were given. Keep it clear and short. "
        "If the web disagreed with the notes, point that out. Do not call any tool. Reply in plain text only."
    ),
    agent=writer,
    context=[research_task, verify_task],
    expected_output="A clear final answer of one to three short paragraphs.",
)
print("Agents and tasks ready")
```

```python
import asyncio, json, textwrap

MEM_FILE = "memory.json"
history = json.load(open(MEM_FILE)) if os.path.exists(MEM_FILE) else []

async def ask(question):
    # Stage 1: Researcher, then Verifier
    stage1 = Crew(agents=[researcher, verifier],
                  tasks=[research_task, verify_task],
                  process=Process.sequential)
    await asyncio.to_thread(stage1.kickoff, inputs={"question": question})
    await asyncio.sleep(20)   # breathing room for Groq's per-minute token limit

    # Stage 2: Writer, with earlier questions and answers as memory
    past = "\n".join(f"Q: {h['q']}\nA: {h['a'][:300]}" for h in history[-3:])
    full = question if not past else f"{question}\n\n(Earlier questions and answers, use if relevant:\n{past})"
    stage2 = Crew(agents=[writer], tasks=[write_task], process=Process.sequential)
    res = await asyncio.to_thread(stage2.kickoff, inputs={"question": full})

    history.append({"q": question, "a": res.raw})
    json.dump(history, open(MEM_FILE, "w"))
    return res.raw

try:
    print(textwrap.fill(await ask("What are the main purposes of customer relationship management?"), 100))
    print("\nWaiting for 60 seconds to avoid Groq Rate Limits...\n")
    await asyncio.sleep(60)
    print(textwrap.fill(await ask("How does implementing CRM help achieve those purposes?"), 100))
except Exception as e:
    print(f"An error occurred: {type(e).__name__} - {e}")
```
